```python
from langchain_community.tools import tool
@tool
def multiply(a :int, b:int) -> int:
    """Multiply two numbers."""
    return a * b
result = multiply.invoke({"a": 5, "b": 3})
print(result)
```

```python
@tool
def add(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b
class Math_toolkit:
    def get_tools(self):
        return [multiply, add]
```

```python
toolkit = Math_toolkit()
tools = toolkit.get_tools()
print(tools[0].name)
print(tools[1].name)
```
