```python
from langchain_chroma import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.documents import Document
from dotenv import load_dotenv
load_dotenv()
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)
```

```python
doc1 = Document(
    page_content="The Hubble Space Telescope has captured some of the most detailed images of distant galaxies, helping astronomers understand the evolution of the universe.",
    metadata={"topic": "Astronomy"}
)

doc2 = Document(
    page_content="NASA's Artemis program aims to return humans to the Moon and establish a sustainable lunar presence as a stepping stone for future Mars missions.",
    metadata={"topic": "Moon Mission"}
)

doc3 = Document(
    page_content="Mars is often called the Red Planet because of its iron oxide-rich surface. Scientists are studying it to determine whether it once supported microbial life.",
    metadata={"topic": "Mars"}
)

doc4 = Document(
    page_content="The James Webb Space Telescope observes the universe in infrared light, allowing it to detect the earliest galaxies and study the atmospheres of exoplanets.",
    metadata={"topic": "Space Telescope"}
)

doc5 = Document(
    page_content="SpaceX's reusable Falcon 9 rocket has significantly reduced the cost of launching satellites by landing and reusing its first-stage booster.",
    metadata={"topic": "Rocket Technology"}
)
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)
docs = [doc1, doc2, doc3, doc4, doc5]
vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory="my_chroma_db",
    collection_name="sample"
)
vector_store.add_documents(docs)
vector_store.get(
    include=["metadatas", "documents", "embeddings"]
)
```

```python
vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory="my_chroma_db",
    collection_name="sample"
)
vector_store.add_documents(docs)
vector_store.get(include=["metadata",'documents', "embeddings"])
```

```python
vector_store.similarity_search(
    query="Tell me about the Hubble Space Telescope.", k=2)
```

```python
vector_store.similarity_search_with_score(
    query="Tell me about the Hubble Space Telescope.", k=2)
```
